```python
# Write a program to:
# 1. Print every student name using a loop over the dictionary.
# 2. Print every mark using .values().
# 3. Print each student and mark using .items().
# 4. Create a new dictionary containing only students who scored at least 50.
# 5. Create a dictionary mapping each student's name to their marks plus 5 bonus marks.
# 6. Create a dictionary containing only students who scored at least 80.
marks = {
    "Amit": 75,
    "Neha": 92,
    "Rahul": 48,
    "Priya": 85,
    "Sneha": 39
}
for key in marks:
    print(key)

for mark in marks.values():
    print(mark)

for name, mark in marks.items():
    print(name,mark)

above_50 = { name: mark for name, mark in marks.items() if mark>=50}
print(above_50)

bonus_5 = { name:mark+5 for name,mark in marks.items() }
print(bonus_5)

above_80 = { name: mark for name,mark  in marks.items() if mark>=80 }
print(above_80)
```
